```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load real-world dataset
df = sns.load_dataset("iris")

# Display first records
print("First 5 records:")
print(df.head())

# Display last records
print("\nLast 5 records:")
print(df.tail())

# Dataset information
print("\nDataset information:")
print(df.info())

# Summary statistics
print("\nSummary statistics:")
print(df.describe())

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Drop missing values if present
df_clean = df.dropna()

print("\nShape after removing missing values:")
print(df_clean.shape)

# Correlation matrix
correlation = df_clean.select_dtypes(
    include="number"
).corr()

print("\nCorrelation Matrix:")
print(correlation)

# Heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Matrix - Iris Dataset")
plt.show()

# Histogram
plt.figure(figsize=(8, 5))
sns.histplot(
    data=df_clean,
    x="sepal_length",
    hue="species",
    kde=True
)

plt.title("Sepal Length Distribution")
plt.show()

# Scatter plot
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df_clean,
    x="sepal_length",
    y="petal_length",
    hue="species",
    style="species"
)

plt.title("Sepal Length vs Petal Length")
plt.show()

# Box plot
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df_clean,
    x="species",
    y="petal_length"
)

plt.title("Petal Length by Species")
plt.show()
```
