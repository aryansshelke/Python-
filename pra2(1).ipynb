```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50])
print("1D Array:", arr1)
```

```python
print("First element:", arr1[0])
print("Last element:", arr1[-1])
```

```python
print("Elements from index 1 to 3:", arr1[1:4])
```

```python
arr2 = np.arange(1, 11)
print("\nArray using arange:", arr2)
```

```python
arr3 = np.linspace(0, 10, 5)
print("Array using linspace:", arr3)
```

```python
arr2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print("\n2D Array:\n", arr2d)
```

```python
print("Element at row 0, column 1:", arr2d[0, 1])
print("Element at row 1, column 2:", arr2d[1, 2])
```

```python
print("First row:", arr2d[0, :])
print("First column:", arr2d[:, 0])
print("First two columns:\n", arr2d[:, 0:2])
```

```python
arr4 = np.arange(1, 13)
reshaped = np.reshape(arr4, (3, 4))
print(reshaped)
```

```python
print("\nSum:", np.sum(arr2d))
print("Mean:", np.mean(arr2d))
```
