```python
import numpy as np
#vectorization
a = np.array([1, 4, 9]) 
b = np.array([10, 20, 30])
```

```python
print("Addition:", np.add(a, b))
```

```python
#universal functions
print("Square root:", np.sqrt(a)) 
print("Exponential:", np.exp([1, 2, 3]))
```

```python
#Broadcasting
print("Broadcasting:", a + 2)
```
