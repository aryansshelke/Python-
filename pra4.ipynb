```python
import numpy as np 
# Random numbers between 0 and 1
a = np.random.rand(5) 
print("Random:", a)
```

```python
# Standard normal distribution 
b = np.random.randn(5) 
print("Normal:", b)
```

```python
# Random integers 
c = np.random.randint(1, 10, 5) 
print("Integers:", c)
```

```python
#Normal distribution with mean=50, std=10 
d = np.random.normal(50, 10, 5) 
print("Normal distribution:", d)
```

```python
# Statistics 
print("Mean:", np.mean(d)) 
print("Standard deviation:", np.std(d))
```
