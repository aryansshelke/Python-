```python
import pandas as pd

s=pd.S

# Create a DataFrame
data = {
    "Name": ["Amit", "Riya", "John", "Sara"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 92, 78, 88]
}

df = pd.DataFrame(data)

# Display data
print(df)

# First rows
print(df.head())

# Last rows
print(df.tail())

# Statistical summary
print(df.describe())

# Select a column
print(df["Marks"])

# Filter data
print(df[df["Marks"] > 80])

# Calculate average
print(df["Marks"].mean())

# Export to CSV
df.to_csv("students.csv", index=False)

# Import from CSV
new_df = pd.read_csv("students.csv")

print(new_df)
```
