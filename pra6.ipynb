```python
import pandas as pd

data = {
    "Name": ["Amit", "Riya", "John", "Sara", "Raj"],
    "Department": ["IT", "HR", "IT", "HR", "Sales"],
    "Age": [20, None, 19, 22, 21],
    "Marks": [85, 92, None, 88, 75]
}

df = pd.DataFrame(data)

print("orginal dataframe")
print(df)

# Check missing values
print("Check missing values:")
print(df.isnull().sum())

# Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
print("\n After replacing missing values:")
print(df)


# Filter rows
print("\n students having marks greater than 80:")
print(df[df["Marks"] > 80])

# Select columns
print("\n Selected columns:")
print(df[["Name", "Marks"]])

# Sort
print("\n sorted by marks:")
print(df.sort_values("Marks", ascending=False))

# Group
print("\n Average marks by Department:")
print(df.groupby("Department")["Marks"].mean())

#Drop missing values
print("\n Data after dropna():")
print(df.dropna())
```
