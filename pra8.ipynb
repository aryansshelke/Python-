```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 4, 9, 16, 25]

plt.figure(figsize=(12, 8))

# Plot 1
plt.subplot(2, 2, 1)
plt.plot(
    x, y1,
    color="blue",
    marker="o",
    linestyle="-",
    label="2x"
)
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)

# Plot 2
plt.subplot(2, 2, 2)
plt.bar(x, y1, color="orange")
plt.title("Bar Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)

# Plot 3
plt.subplot(2, 2, 3)
plt.scatter(x, y2, color="red", label="x²")
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)

# Plot 4
plt.subplot(2, 2, 4)
plt.plot(
    x, y2,
    color="green",
    marker="s",
    label="x²"
)
plt.title("Styled Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()

```
