```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Dataset
data = {
    "Name": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Marks": [65, 72, 80, 85, 90, 75, 88, 95],
    "StudyHours": [2, 3, 4, 5, 6, 3, 5, 7]
}

df = pd.DataFrame(data)

# Line plot
plt.figure(figsize=(8, 5))
sns.lineplot(
    data=df,
    x="StudyHours",
    y="Marks",
    marker="o"
)
plt.title("Study Hours vs Marks")
plt.show()

# Bar plot
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x="Name",
    y="Marks"
)
plt.title("Student Marks")
plt.show()

# Histogram
plt.figure(figsize=(8, 5))
sns.histplot(
    data=df,
    x="Marks",
    bins=5,
    kde=True
)
plt.title("Marks Distribution")
plt.show()

# Box plot
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    y="Marks"
)
plt.title("Marks Box Plot")
plt.show()

# Scatter plot
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="StudyHours",
    y="Marks"
)
plt.title("Study Hours vs Marks")
plt.show()
```
