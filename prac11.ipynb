```python
import numpy as np 

A=np.array([[2,1],
             [1,3]])

B=np.array([5,7])

print("Matrix A:")
print(A)

X=np.linalg.solve(A,B)
print("\nsolution of A:")
print(np.linalg.inv(A))

eigenvalues,eigenvectors=np.linalg.eig(A)
print("\nEigenvalues:")
print(eigenvectors)

print("\nEigenvectors:")
print(eigenvectors)

print("\nTranspose:")
print(A.T)
```
